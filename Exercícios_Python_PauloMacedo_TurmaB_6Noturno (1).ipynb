**Enunciado:** Conte o número de ocorrências de cada caractere em uma string fornecida.

```python
texto = input("Digite uma frase: ")

contagem = {}  # dicionário: chave = caractere, valor = quantidade de ocorrências

for caractere in texto:
    # get(caractere, 0) devolve o valor atual ou 0 caso o caractere ainda não exista
    contagem[caractere] = contagem.get(caractere, 0) + 1

for caractere, quantidade in contagem.items():
    print(f"'{caractere}': {quantidade}")
```

## Exercício 18: Número Primo

**Enunciado:** Crie um programa que verifique se um número fornecido é primo.

**Como resolvi:** Números menores que 2 não são primos. Para os demais, testo divisores apenas até a raiz quadrada do número, pois qualquer divisor maior teria um par menor que a raiz. Ao encontrar um divisor, interrompo o laço com `break`.

```python
numero = int(input("Digite um número inteiro: "))

eh_primo = numero >= 2  # números menores que 2 não são primos

# basta testar divisores até a raiz quadrada do número,
# pois divisores maiores sempre têm um par menor que a raiz
for i in range(2, int(numero ** 0.5) + 1):
    if numero % i == 0:
        eh_primo = False
        break  # encontrou um divisor, não precisa continuar

print(f"{numero} é primo" if eh_primo else f"{numero} não é primo")
```

## Exercício 19: Jogo de Adivinhação

**Enunciado:** Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.

**Como resolvi:** Uso `random.randint(1, 100)` para sortear o número secreto e um laço `while` que só termina quando o palpite estiver correto, dando dicas de maior ou menor e contando as tentativas.

```python
import random  # módulo para gerar números aleatórios

numero_secreto = random.randint(1, 100)  # sorteia um inteiro entre 1 e 100
tentativas = 0
acertou = False

print("Adivinhe o número entre 1 e 100!")

while not acertou:
    palpite = int(input("Seu palpite: "))
    tentativas += 1

    if palpite < numero_secreto:
        print("O número secreto é MAIOR. Tente de novo!")
    elif palpite > numero_secreto:
        print("O número secreto é MENOR. Tente de novo!")
    else:
        acertou = True
        print(f"Parabéns! Você acertou em {tentativas} tentativa(s).")
```

## Exercício 20: Matriz Transposta

**Enunciado:** Dada uma matriz (lista de listas), crie sua matriz transposta.

**Como resolvi:** Na transposta, as linhas viram colunas, ou seja, o elemento da posição [i][j] vai para [j][i]. Uso uma list comprehension aninhada que monta cada nova linha a partir das colunas da matriz original. A função `zip(*matriz)` seria uma alternativa pronta.

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6]
]  # matriz 2 x 3 (2 linhas e 3 colunas)

# na transposta, as linhas viram colunas: o elemento [i][j] vai para [j][i]
transposta = [[linha[i] for linha in matriz] for i in range(len(matriz[0]))]

print("Matriz original:")
for linha in matriz:
    print(linha)

print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```

## Exercício 21: Busca Binária

**Enunciado:** Implemente a busca binária para encontrar um elemento em uma lista ordenada.

**Como resolvi:** Comparo o alvo com o elemento central da lista. Se for igual, encontrei; se o central for menor, descarto a metade da esquerda; se for maior, descarto a da direita. O espaço de busca cai pela metade a cada passo, o que torna o algoritmo muito eficiente (complexidade O(log n)).

```python
def busca_binaria(lista, alvo):
    """Devolve o índice de alvo em lista (ordenada) ou -1 se não encontrar."""
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2  # índice central do trecho atual

        if lista[meio] == alvo:
            return meio           # encontrou o elemento
        elif lista[meio] < alvo:
            inicio = meio + 1     # descarta a metade da esquerda
        else:
            fim = meio - 1        # descarta a metade da direita

    return -1  # percorreu tudo sem encontrar

numeros = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(f"Lista: {numeros}")
print(f"Índice do 23: {busca_binaria(numeros, 23)}")
print(f"Índice do 91: {busca_binaria(numeros, 91)}")
print(f"Índice do 7 (não existe): {busca_binaria(numeros, 7)}")
```

## Exercício 22: Criptografia de Cifra de César

**Enunciado:** Crie uma função para criptografar uma mensagem usando a cifra de César.

**Como resolvi:** Cada letra é convertida em número com `ord()`, deslocada e convertida de volta com `chr()`. O operador `% 26` faz o alfabeto dar a volta (depois do z vem o a) e a base muda conforme a letra seja maiúscula ou minúscula. Caracteres que não são letras permanecem inalterados.

```python
def cifrar_cesar(mensagem, deslocamento):
    """Desloca cada letra da mensagem 'deslocamento' posições no alfabeto."""
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():  # apenas letras são cifradas
            # a base define se trabalhamos com maiúsculas ou minúsculas na tabela ASCII
            base = ord('A') if caractere.isupper() else ord('a')
            # ord() converte a letra em número, aplicamos o deslocamento
            # e o % 26 garante que o alfabeto "dê a volta" (depois do z vem o a)
            nova_posicao = (ord(caractere) - base + deslocamento) % 26
            resultado += chr(base + nova_posicao)  # chr() converte de volta para letra
        else:
            resultado += caractere  # espaços e pontuação ficam como estão

    return resultado

mensagem = "Ataque ao amanhecer!"
cifrada = cifrar_cesar(mensagem, 3)
print(f"Mensagem original: {mensagem}")
print(f"Mensagem cifrada:  {cifrada}")
```

## Exercício 24: Validação de CPF

**Enunciado:** Implemente uma função para validar um CPF brasileiro.

**Como resolvi:** Sigo o algoritmo oficial dos dígitos verificadores. Limpo a entrada mantendo só os números, rejeito tamanhos diferentes de 11 e sequências de dígitos repetidos, calculo o primeiro dígito com pesos de 10 a 2 e o segundo com pesos de 11 a 2, comparando com os dígitos informados. O CPF 111.444.777-35 é um exemplo de teste clássico, válido pelo cálculo.

```python
def validar_cpf(cpf):
    """Valida um CPF brasileiro usando os dois dígitos verificadores."""
    # mantém apenas os números, permitindo entradas como 111.444.777-35
    cpf = "".join(c for c in cpf if c.isdigit())

    # um CPF válido tem exatamente 11 dígitos
    if len(cpf) != 11:
        return False

    # CPFs com todos os dígitos iguais (ex.: 111.111.111-11) passam no cálculo,
    # mas são inválidos por definição
    if cpf == cpf[0] * 11:
        return False

    # primeiro dígito verificador: pesos de 10 a 2 sobre os 9 primeiros dígitos
    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
    resto = soma % 11
    digito1 = 0 if resto < 2 else 11 - resto
    if digito1 != int(cpf[9]):
        return False

    # segundo dígito verificador: pesos de 11 a 2 sobre os 10 primeiros dígitos
    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
    resto = soma % 11
    digito2 = 0 if resto < 2 else 11 - resto

    return digito2 == int(cpf[10])

# Testes com um CPF de exemplo válido e dois inválidos
print(validar_cpf("111.444.777-35"))  # True, CPF de teste válido pelo cálculo
print(validar_cpf("111.111.111-11"))  # False, dígitos repetidos
print(validar_cpf("123.456.789-00"))  # False, dígitos verificadores errados
```

## Exercício 25: Classificação de Números

**Enunciado:** Classifique uma lista de números em três categorias: positivos, negativos e zeros.

```python
numeros = [5, -3, 0, 12, -7, 0, 8, -1]

positivos = []
negativos = []
zeros = []

for numero in numeros:
    if numero > 0:
        positivos.append(numero)
    elif numero < 0:
        negativos.append(numero)
    else:
        zeros.append(numero)

print(f"Lista analisada: {numeros}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Zeros: {zeros}")
```

## Exercício 27: Anagramas

**Enunciado:** Verifique se duas palavras são anagramas (ou seja, se são formadas pelas mesmas letras).

**Como resolvi:** Se duas palavras usam exatamente as mesmas letras, ao ordenar essas letras o resultado é igual. Então padronizo (minúsculas, sem espaços) e comparo `sorted(p1) == sorted(p2)`.

```python
def sao_anagramas(palavra1, palavra2):
    """Duas palavras são anagramas quando usam exatamente as mesmas letras."""
    # padroniza: minúsculas e sem espaços
    p1 = palavra1.lower().replace(" ", "")
    p2 = palavra2.lower().replace(" ", "")
    # se as letras ordenadas forem iguais, as palavras têm as mesmas letras
    return sorted(p1) == sorted(p2)

print(sao_anagramas("amor", "roma"))     # True
print(sao_anagramas("banana", "abanan")) # True
print(sao_anagramas("python", "java"))   # False
```

## Exercício 30: Jogo da Forca

**Enunciado:** Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

**Como resolvi:** Sorteio uma palavra com `random.choice()` e uso dois conjuntos (`set`) para letras certas e erradas, o que evita repetições automaticamente. A cada rodada exibo a palavra com `_` nas letras não descobertas, valido a entrada e controlo as 6 tentativas. O `else` do `while` trata a derrota, pois só executa quando o laço termina sem `break`.

```python
import random

palavras = ["python", "programa", "computador", "teclado", "algoritmo"]
palavra_secreta = random.choice(palavras)  # sorteia uma palavra da lista

letras_descobertas = set()  # conjuntos evitam letras repetidas automaticamente
letras_erradas = set()
tentativas_restantes = 6

print("Jogo da Forca! Adivinhe a palavra.")

while tentativas_restantes > 0:
    # monta a exibição: letras descobertas aparecem, as demais viram _
    exibicao = " ".join(letra if letra in letras_descobertas else "_"
                        for letra in palavra_secreta)
    print(f"\nPalavra: {exibicao}")
    print(f"Tentativas restantes: {tentativas_restantes}")
    print(f"Letras erradas: {sorted(letras_erradas)}")

    # vitória: todas as letras da palavra já foram descobertas
    if set(palavra_secreta) <= letras_descobertas:
        print("Parabéns, você venceu!")
        break

    letra = input("Digite uma letra: ").lower()

    if len(letra) != 1 or not letra.isalpha():
        print("Digite apenas UMA letra válida.")
        continue

    if letra in letras_descobertas or letra in letras_erradas:
        print("Você já tentou essa letra.")
        continue

    if letra in palavra_secreta:
        letras_descobertas.add(letra)
        print("Boa! A letra existe na palavra.")
    else:
        letras_erradas.add(letra)
        tentativas_restantes -= 1
        print("Essa letra não existe na palavra.")
else:
    # o else do while executa quando o laço termina sem break, ou seja, sem vitória
    print(f"\nFim de jogo! A palavra era '{palavra_secreta}'.")
```

## Exercício 31: Ordenação por Bolha

**Enunciado:** Implemente o algoritmo de ordenação por bolha (Bubble Sort).

**Como resolvi:** Comparo pares vizinhos e troco quando estão fora de ordem; a cada passada, o maior elemento restante "borbulha" para o fim, por isso o limite do laço interno diminui. Incluo a otimização de parar quando uma passada inteira acontece sem trocas, sinal de que a lista já está ordenada.

```python
def bubble_sort(lista):
    """Ordena comparando pares vizinhos e 'empurrando' o maior para o fim."""
    n = len(lista)
    lista = lista.copy()  # trabalha em uma cópia para preservar a lista original

    for i in range(n - 1):
        trocou = False
        # a cada passada, o maior elemento restante vai para o final,
        # por isso o limite diminui: n - 1 - i
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]  # troca os vizinhos
                trocou = True
        if not trocou:
            break  # otimização: se não houve troca, a lista já está ordenada

    return lista

numeros = [64, 34, 25, 12, 22, 11, 90]
print(f"Original: {numeros}")
print(f"Ordenada: {bubble_sort(numeros)}")
```

## Exercício 32: Jogo da Velha

**Enunciado:** Crie um jogo da velha para dois jogadores.

**Como resolvi:** Represento o tabuleiro como uma lista de 9 posições e organizo o código em funções: uma para desenhar o tabuleiro e outra para conferir as 8 combinações vencedoras (linhas, colunas e diagonais). O laço principal alterna os jogadores, valida as jogadas (posição de 1 a 9 e casa livre) e detecta vitória ou empate.

```python
def exibir_tabuleiro(tabuleiro):
    """Mostra o tabuleiro 3 x 3 na tela."""
    print()
    for i in range(0, 9, 3):
        print(f" {tabuleiro[i]} | {tabuleiro[i + 1]} | {tabuleiro[i + 2]} ")
        if i < 6:
            print("---+---+---")
    print()

def verificar_vitoria(tabuleiro, simbolo):
    """Confere todas as combinações vencedoras (linhas, colunas e diagonais)."""
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),  # linhas
        (0, 3, 6), (1, 4, 7), (2, 5, 8),  # colunas
        (0, 4, 8), (2, 4, 6)              # diagonais
    ]
    return any(tabuleiro[a] == tabuleiro[b] == tabuleiro[c] == simbolo
               for a, b, c in combinacoes)

# o tabuleiro começa com os números das posições para orientar os jogadores
tabuleiro = [str(i) for i in range(1, 10)]
jogador_atual = "X"
jogadas = 0

print("Jogo da Velha! Jogadores X e O escolhem posições de 1 a 9.")

while True:
    exibir_tabuleiro(tabuleiro)
    escolha = input(f"Jogador {jogador_atual}, escolha uma posição (1 a 9): ")

    # validação: precisa ser um número de 1 a 9
    if not escolha.isdigit() or not 1 <= int(escolha) <= 9:
        print("Posição inválida! Digite um número de 1 a 9.")
        continue

    posicao = int(escolha) - 1  # converte para índice da lista (0 a 8)

    # validação: a casa escolhida precisa estar livre
    if tabuleiro[posicao] in "XO":
        print("Essa casa já está ocupada, escolha outra.")
        continue

    tabuleiro[posicao] = jogador_atual
    jogadas += 1

    if verificar_vitoria(tabuleiro, jogador_atual):
        exibir_tabuleiro(tabuleiro)
        print(f"O jogador {jogador_atual} venceu! Parabéns!")
        break

    if jogadas == 9:  # todas as casas preenchidas sem vencedor
        exibir_tabuleiro(tabuleiro)
        print("Deu velha! Empate.")
        break

    # alterna o jogador da vez
    jogador_atual = "O" if jogador_atual == "X" else "X"
```

## Exercício 33: Verificação de Anagramas

**Enunciado:** Verifique se duas palavras fornecidas são anagramas uma da outra.

**Como resolvi:** Este exercício repete a ideia do 27, então aproveito para mostrar outra abordagem: o `Counter` do módulo `collections` conta as letras automaticamente, e anagramas produzem contagens idênticas. Aqui as palavras vêm do usuário via `input()`.

```python
from collections import Counter  # Counter conta ocorrências automaticamente

palavra1 = input("Digite a primeira palavra: ").lower().replace(" ", "")
palavra2 = input("Digite a segunda palavra: ").lower().replace(" ", "")

# Counter cria um dicionário {letra: quantidade}; anagramas geram contagens iguais
if Counter(palavra1) == Counter(palavra2):
    print(f"'{palavra1}' e '{palavra2}' SÃO anagramas!")
else:
    print(f"'{palavra1}' e '{palavra2}' NÃO são anagramas.")
```

## Exercício 37: Funções Recursivas

**Enunciado:** Implemente uma função recursiva para calcular o fatorial de um número.

**Como resolvi:** A recursão precisa de um caso base, que interrompe as chamadas (n menor ou igual a 1 devolve 1), e do caso recursivo, que reduz o problema: n! = n * (n - 1)!. Cada chamada resolve uma versão menor do mesmo problema até chegar ao caso base.

```python
def fatorial(n):
    """Calcula n! chamando a própria função com um problema menor."""
    if n <= 1:          # caso base: 0! = 1 e 1! = 1, interrompe a recursão
        return 1
    return n * fatorial(n - 1)  # caso recursivo: n! = n * (n - 1)!

numero = int(input("Digite um número para calcular o fatorial: "))

if numero < 0:
    print("Não existe fatorial de número negativo.")
else:
    print(f"{numero}! = {fatorial(numero)}")
```

## Exercício 40: Lista de Compras

**Enunciado:** Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

**Como resolvi:** Um laço `while True` mantém um menu em execução até o usuário escolher sair. Uso os métodos de lista `append()` e `remove()`, verifico se o item existe antes de remover e uso `enumerate()` para listar os itens com numeração.

```python
lista_compras = []

while True:
    print("\n===== LISTA DE COMPRAS =====")
    print("1. Adicionar item")
    print("2. Remover item")
    print("3. Listar itens")
    print("4. Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        item = input("Qual item deseja adicionar? ").strip()
        if item:
            lista_compras.append(item)
            print(f"'{item}' adicionado à lista.")
        else:
            print("Item vazio não pode ser adicionado.")

    elif opcao == "2":
        item = input("Qual item deseja remover? ").strip()
        if item in lista_compras:
            lista_compras.remove(item)
            print(f"'{item}' removido da lista.")
        else:
            print(f"'{item}' não está na lista.")

    elif opcao == "3":
        if lista_compras:
            print("Itens na lista:")
            for i, item in enumerate(lista_compras, start=1):
                print(f"{i}. {item}")
        else:
            print("A lista está vazia.")

    elif opcao == "4":
        print("Encerrando o programa. Boas compras!")
        break

    else:
        print("Opção inválida, tente novamente.")
```

## Exercício 42: Números Primos até N

**Enunciado:** Gere todos os números primos até um número n fornecido.

**Como resolvi:** Encapsulo o teste de primalidade do exercício 18 em uma função `eh_primo()` e uso uma list comprehension com filtro para reunir todos os primos de 2 até n.

```python
def eh_primo(numero):
    """Devolve True se o número for primo."""
    if numero < 2:
        return False
    for i in range(2, int(numero ** 0.5) + 1):  # testa divisores até a raiz quadrada
        if numero % i == 0:
            return False
    return True

n = int(input("Gerar números primos até qual valor? "))

# list comprehension com filtro: guarda apenas os números aprovados por eh_primo()
primos = [numero for numero in range(2, n + 1) if eh_primo(numero)]

print(f"Primos até {n}: {primos}")
```

## Exercício 46: Calcular Média e Desvio Padrão

**Enunciado:** Calcule a média e o desvio padrão de uma lista de números.

**Como resolvi:** Calculo a média, depois a variância populacional (média dos quadrados das distâncias até a média) e o desvio padrão como a raiz quadrada da variância. No final, confiro o resultado com o módulo `statistics` da biblioteca padrão, cuja função `pstdev()` calcula o mesmo desvio padrão populacional.

```python
import statistics  # módulo da biblioteca padrão, usado para conferir o resultado

numeros = [10, 12, 23, 23, 16, 23, 21, 16]

# média: soma dos valores dividida pela quantidade
media = sum(numeros) / len(numeros)

# variância populacional: média dos quadrados das distâncias até a média
variancia = sum((x - media) ** 2 for x in numeros) / len(numeros)

# desvio padrão: raiz quadrada da variância
desvio_padrao = variancia ** 0.5

print(f"Lista: {numeros}")
print(f"Média: {media:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")

# conferindo com o módulo statistics (pstdev = desvio padrão populacional)
print(f"Conferência: média = {statistics.mean(numeros):.2f}, "
      f"desvio = {statistics.pstdev(numeros):.2f}")
```

## Exercício 48: Ordenação por Inserção

**Enunciado:** Implemente o algoritmo de ordenação por inserção (Insertion Sort).

**Como resolvi:** Percorro a lista a partir do segundo elemento e insiro cada "chave" na posição correta da parte já ordenada, deslocando para a direita os elementos maiores que ela. É o mesmo movimento de quem organiza cartas na mão.

```python
def insertion_sort(lista):
    """Ordena inserindo cada elemento na posição correta da parte já ordenada."""
    lista = lista.copy()  # preserva a lista original

    for i in range(1, len(lista)):
        chave = lista[i]  # elemento a ser posicionado
        j = i - 1
        # desloca para a direita os elementos maiores que a chave
        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = chave  # insere a chave no espaço aberto

    return lista

numeros = [12, 11, 13, 5, 6, 2]
print(f"Original: {numeros}")
print(f"Ordenada: {insertion_sort(numeros)}")
```

## Exercício 49: Soma dos Dígitos

**Enunciado:** Calcule a soma dos dígitos de um número inteiro fornecido.

**Como resolvi:** Uso aritmética: o resto da divisão por 10 (`% 10`) isola o último dígito e a divisão inteira (`// 10`) o descarta, repetindo até acabarem os dígitos. `abs()` trata números negativos. Uma alternativa seria converter o número para string e somar os dígitos com `int()`.

```python
numero = int(input("Digite um número inteiro: "))

n = abs(numero)  # abs() ignora o sinal, útil para números negativos
soma = 0

while n > 0:
    soma += n % 10  # o resto da divisão por 10 isola o último dígito
    n //= 10        # a divisão inteira por 10 descarta o último dígito

# observação: se o número for 0, o laço não executa e a soma permanece 0, que é o correto
print(f"A soma dos dígitos de {numero} é {soma}")

# Alternativa usando string: sum(int(d) for d in str(abs(numero)))
```

## Exercício 50: Cálculo de Frequência de Letras

**Enunciado:** Calcule a frequência de cada letra em uma string fornecida pelo usuário.

**Como resolvi:** Parecido com o exercício 12, mas aqui filtro com `isalpha()` para considerar apenas letras, ignorando espaços, números e pontuação, e exibo o resultado em ordem alfabética com `sorted()` para facilitar a leitura.

```python
texto = input("Digite um texto: ").lower()

frequencia = {}

for caractere in texto:
    if caractere.isalpha():  # considera apenas letras, ignorando espaços e símbolos
        frequencia[caractere] = frequencia.get(caractere, 0) + 1

print("Frequência de cada letra:")
for letra in sorted(frequencia):  # ordem alfabética para facilitar a leitura
    print(f"{letra}: {frequencia[letra]}")
```
